```python
import pandas as pd
#percentage of null values
null_percent = df.isnull().mean() * 100
```

```python
df['saledate'] = df['saledate'].ffill()
```

```python
for col in df.columns:
    sellingprice = df[col].isnull().mean() * 100
    print(f"{col}: {sellingprice:.2f}% missing values")
```

```python
# Count total duplicate rows
duplicate_count = df.duplicated().sum()
print(f"Number of duplicate : {duplicate_count}")
```

```python
# Remove duplicates
df = df.drop_duplicates()

print(f"Shape after removing duplicates: {df.shape}")
```

```python
#Data frames Queries:
1.
# Average, minimum, and maximum car price
avg_price = df['sellingprice'].mean()
min_price = df['sellingprice'].min()
max_price = df['sellingprice'].max()

print(f"Average Car Price: {avg_price:.2f}")
print(f"Minimum Car Price: {min_price}")
print(f"Maximum Car Price: {max_price}")
```

```python
2.
# List all unique car colors
unique_colors = df['color'].unique()
print("Unique car colors:")
print(unique_colors)
```

```python
3.
# Number of unique car brands
unique_brands = df['make'].nunique()

# Number of unique car models
unique_models = df['model'].nunique()

print(f"Number of unique car brands: {unique_brands}")
print(f"Number of unique car models: {unique_models}")
```

```python
4.
# Filter cars with selling price > 165000
high_price_cars = df[df['sellingprice'] > 165000]

print(f"Number of cars with price > 165000: {high_price_cars.shape[0]}")
high_price_cars

```

```python
5.
# Top 5 most frequently sold car models
top_models = df['model'].value_counts().head(5)

print("Top 5 most frequently sold car models:")
print(top_models)
```

```python
6.
# Average selling price grouped by brand (make)
avg_price_by_brand = df.groupby('make')['sellingprice'].mean().sort_values(ascending=False)

print("Average Selling Price by Brand:")
print(avg_price_by_brand)
```

```python
7.
# Minimum selling price grouped by interior type
min_price_by_interior = df.groupby('interior')['sellingprice'].min().sort_values()

print("Minimum Selling Price by Interior:")
print(min_price_by_interior)

```

```python
8.
# Highest odometer reading per year, sorted descending
highest_odometer_per_year = (
    df.groupby('year')['odometer']
      .max()
      .sort_values(ascending=False))

print("Highest Odometer Reading per Year (sorted):")
print(highest_odometer_per_year)
```

```python
9.
# Create a new column 'car_age'
df['car_age'] = 2025 - df['year']

# Preview the first few rows
print(df[['year', 'car_age']].head())
```

```python
10.
# Filter cars with condition >= 48 and odometer > 90000
filtered_cars = df[(df['condition'] >= 48) & (df['odometer'] > 90000)]

# Count them
count_filtered = filtered_cars.shape[0]

print(f"Number of cars with condition >= 48 and odometer > 90000: {count_filtered}")
```

```python
11.
# Filter dataset for newer cars (year > 2013)
newer_cars = df[df['year'] > 2013]

# Group by state and calculate average selling price
avg_price_by_state = (
    newer_cars.groupby('state')['sellingprice']
    .mean()
    .sort_values(ascending=False))

print("Average selling price of newer cars (year > 2013) by state:")
print(avg_price_by_state.head(10))  # top 10 states
```

```python
12.
# Step 1: Calculate the threshold for top 20% condition
condition_threshold = df['condition'].quantile(0.80)

# Step 2: Filter cars with condition >= threshold
excellent_cars = df[df['condition'] >= condition_threshold]

# Step 3: Group by make and calculate average selling price
avg_price_by_make = (
    excellent_cars.groupby('make')['sellingprice']
    .mean()
    .sort_values(ascending=True)   # lowest first
)

# Step 4: Show the top 10 makes with lowest average price
print("Makes with lowest average price among excellent condition cars:")
print(avg_price_by_make.head(10))
```

```python
2.
import matplotlib.pyplot as plt

# Group by year and calculate average selling price
avg_price_by_year = df.groupby('year')['sellingprice'].mean().sort_index()

# Plot as a bar chart
plt.figure(figsize=(10,6))
avg_price_by_year.plot(kind='bar', color='skyblue', edgecolor='black')

plt.title("Average Selling Price by Year")
plt.xlabel("Year")
plt.ylabel("Average Selling Price")
plt.xticks(rotation=45)
plt.show()
```

```python
3.
import matplotlib.pyplot as plt
import pandas as pd

# Group odometer into bins (e.g., every 20,000 miles/km)
odometer_bins = pd.cut(df['odometer'], bins=10)
avg_price_by_odometer = df.groupby(odometer_bins)['sellingprice'].mean()

# Plot scatter/line chart
plt.figure(figsize=(10,6))
avg_price_by_odometer.plot(marker='o', linestyle='-', color='teal')

plt.title("Average Selling Price by Odometer")
plt.xlabel("Odometer Range")
plt.ylabel("Average Selling Price")
plt.xticks(rotation=45)
plt.show()
```

```python
4.
import matplotlib.pyplot as plt

# Count number of cars sold per state
cars_sold_by_state = df['state'].value_counts()

# Plot bar chart
plt.figure(figsize=(12,6))
cars_sold_by_state.plot(kind='bar', color='skyblue', edgecolor='black')

plt.title("Number of Cars Sold per State")
plt.xlabel("State")
plt.ylabel("Number of Cars Sold")
plt.xticks(rotation=90)
plt.show()

# Top 3 states
top3_states = cars_sold_by_state.head(3)
print("Top 3 highest car selling states:")
print(top3_states)
```

```python
5.
import matplotlib.pyplot as plt

# Count number of cars sold per state
cars_sold_by_state = df['state'].value_counts()

# Plot bar chart
plt.figure(figsize=(12,6))
cars_sold_by_state.plot(kind='bar', color='skyblue', edgecolor='black')

plt.title("Number of Cars Sold per State")
plt.xlabel("State")
plt.ylabel("Number of Cars Sold")
plt.xticks(rotation=90)
plt.show()

# Top 3 states
top3_states = cars_sold_by_state.head(3)
print("Top 3 highest car selling states:")
print(top3_states)
```

```python
6.
import pandas as pd
import matplotlib.pyplot as plt

# Create bins of size 10 for condition scores
bins = range(0, int(df['condition'].max()) + 10, 10)
labels = [f"{b}-{b+9}" for b in bins[:-1]]

# Assign condition ranges
df['condition_range'] = pd.cut(df['condition'], bins=bins, labels=labels, right=False)

# Count number of cars in each condition range
cars_by_condition = df['condition_range'].value_counts().sort_index()

# Plot bar chart
plt.figure(figsize=(10,6))
cars_by_condition.plot(kind='bar', color='orange', edgecolor='black')

plt.title("Number of Cars Sold by Condition Score Ranges (size 10)")
plt.xlabel("Condition Score Range")
plt.ylabel("Number of Cars Sold")
plt.xticks(rotation=45)
plt.show()
```

```python
7.
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Initial box plot of selling prices grouped by color
plt.figure(figsize=(12,6))
sns.boxplot(x='color', y='sellingprice', data=df)
plt.title("Distribution of Car Selling Prices by Color (with outliers)")
plt.xticks(rotation=45)
plt.show()

# --- Remove outliers using IQR method ---
Q1 = df['sellingprice'].quantile(0.25)
Q3 = df['sellingprice'].quantile(0.75)
IQR = Q3 - Q1

# Keep only values within 1.5*IQR
filtered_df = df[(df['sellingprice'] >= Q1 - 1.5*IQR) & (df['sellingprice'] <= Q3 + 1.5*IQR)]

# Box plot again without outliers
plt.figure(figsize=(12,6))
sns.boxplot(x='color', y='sellingprice', data=filtered_df)
plt.title("Distribution of Car Selling Prices by Color (without extreme outliers)")
plt.xticks(rotation=45)
plt.show()
```
